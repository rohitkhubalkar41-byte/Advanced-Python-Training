```python
from abc import ABC, abstractmethod
# from abc import *
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass

  @abstractmethod
  def m2(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 implemented")

  def m2(self):
    print("m2 implemented")

m = MyPython()
m.m1()
m.m2()
```

```python
#If the child class is unable to provide implemention of
#all methods of abstract class then we can declare the child
#class methods with @abstractmethod and complete the remaining
#method

from abc import ABC, abstractmethod
# from abc import *
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass

  @abstractmethod
  def m2(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 implemented")

  @abstractmethod
  def m2(self):
    pass

class Child(MyPython):
  def m2(self):
    print("m2 implemented")

m = Child()
m.m1()
m.m2()
```

```python
#Inside the abstract class we can declare the constructor
#abstract class constructor is executed but object is not created

from abc import ABC, abstractmethod
# from abc import *
class Test(ABC):
  def __init__(self):
    print("Constructor is called...")

  @abstractmethod
  def m1(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 Method implementation")

m=MyPython()
m.m1()
```

```python
class Test1:
  _b=None
  __c=None
  def __init__(self):
    self.a=101
    self._b=201
    self.__c=301

  def show1(self):
    print("a = ",self.a)
    print("_b = ",self._b)
    print("__c = ",self.__c)

class Test2(Test1):
  def show2(self):
    print("a = ",self.a)
    print("_b = ",self._b)
    #print("__c = ",self.__c)

class Test3(Test2):
  def show3(self):
    print("a = ",self.a)
    print("_b = ",self._b)
    #print("__c = ",self.__c)

t1=Test1()
t1.show1()
t2=Test2()
t2.show2()
t3=Test3()
t3.show3()
```

```python
# We can use seek() function to move cursor (file pointer) to
# specified location.

data = "Learning python is very easy for Ashish sir"

f = open("myfile.txt", 'w')
f.write(data)

with open("myfile.txt", 'r+') as f:
    fileData = f.read()
    print(fileData)

    print("Current Cursor Position is", f.tell())

    f.seek(15)
    print("Current Cursor Position is", f.tell())

    f.write("-JAVA-")

    f.seek(0)
    fileData = f.read()
    print(fileData)
```

```python
# WAP to print number of lines, words, characters present in file

import os, sys

fname = input("Enter file name : ")

if os.path.isfile(fname):
    print("File exist :", fname)
    f = open(fname, 'r')
else:
    print("File is not exist :", fname)
    sys.exit(0)

lcount = wcount = ccount = 0

for line in f:
    lcount = lcount + 1
    ccount = ccount + len(line)
    words = line.split()
    wcount = wcount + len(words)

print("Total lines :", lcount)
print("Total words :", wcount)
print("Total characters :", ccount)

f.close()
```
