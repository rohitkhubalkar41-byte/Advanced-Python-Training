```python
import csv

with open("emp.csv", 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(["ENO", "ENAME", "ESALARY", "EADDR"])

    n = int(input("Enter number of employees : "))

    for i in range(n):
        eno = input("Enter employee no : ")
        ename = input("Enter employee name : ")
        esal = input("Enter employee salary : ")
        eaddr = input("Enter employee address : ")

        w.writerow([eno, ename, esal, eaddr])

print("Data inserted successfully!")
```

```python
import numpy as np

students = np.array([
    [85, 90, 78],  # Student 1
    [60, 55, 70],  # Student 2
    [95, 92, 88],  # Student 3
    [40, 50, 45],  # Student 4
    [72, 68, 75]   # Student 5
])

students
```

```python
avg_marks = np.mean(students, axis=1)
print("Average Marks for each student:")
print(avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("Topper Student:",topper)
```

```python
failed_students = np.where(avg_marks < 50)
print("Failed Students:", failed_students[0]+1)
```

```python
#Performance Report
for i, avg in enumerate(avg_marks):
  status = "Pass" if avg >= 50 else "Fail"
  print(f"Student {i+1}: Average Marks = {avg:.2f}")
```

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y)
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
# Bar Chart

months = ['Jan', 'Feb', 'Mar', 'April']
sales = [250, 300, 400, 350]

plt.bar(months, sales, color='skyblue')
plt.title('Monthly sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
plt.barh(months,sales,color='red')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Bar Plot')
plt.show()
```

```python
height = [150,160,170,180]
weight = [50,60,70,80]
plt.scatter(height,weight, color='green')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.title('Scatter Plot')
plt.show()
```

```python
# Pie Chart (Market Share)

import matplotlib.pyplot as plt

brands = ['Brand A', 'Brand B', 'Brand C']
shares = [45, 30, 25]

plt.pie(shares, labels=brands, autopct='%1.1f%%', startangle=90)

plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
#multiple Plot in one figure
x = [1,2,3,4]
y1 = [1,4,9,16]
y2 = [1,2,3,4]
plt.plot(x,y1, label='Square')
plt.plot(x,y2, label='Linear')
plt.legend()
plt.title("Multiple Lines")
plt.show()
```

```python
#Subplots
fig, axs = plt.subplots(2,2)
axs[0,0].plot(x,y1)
axs[0,0].set_title('Square')

axs[0,1].bar(months,sales)
axs[0,1].set_title('Bar')

axs[1,0].scatter(height,weight)
axs[1,0].set_title('Scatter')
```
