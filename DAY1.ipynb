```python
#Accept 5 paper marks, rollno, name;
#and calculate total & percentage
#if percentage >=85, student can take addmission
#else cant take addmission
#using class & object & constructor


class Student:
    def __init__(self):
        self.rollno = int(input("Enter Roll Number: "))
        self.name = input("Enter Student Name: ")

        self.m1 = float(input("Enter Subject 1 Marks: "))
        self.m2 = float(input("Enter Subject 2 Marks: "))
        self.m3 = float(input("Enter Subject 3 Marks: "))
        self.m4 = float(input("Enter Subject 4 Marks: "))
        self.m5 = float(input("Enter Subject 5 Marks: "))

    def calculate(self):
        total = self.m1 + self.m2 + self.m3 + self.m4 + self.m5
        percentage = total / 5

        print("\n----- Student Details -----")
        print("Roll Number:", self.rollno)
        print("Name:", self.name)
        print("Total Marks:", total)
        print("Percentage:", percentage, "%")

        if percentage >= 85:
            print("Eligible for Admission")
        else:
            print("Not Eligible for Admission")


# Creating Object
s1 = Student()
s1.calculate()
```

```python
#Hybrid
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")


b = B()
b.showA()
b.showB()

c = C()
c.showA()
c.showC()
```

```python
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

class D(B, C):
    def showD(self):
        print("showD function")


d = D()
d.showA()
d.showB()
d.showC()
d.showD()
```

```python
#Hierarchical Inheritance
class D:
    def showD(self):
        print("showD function")

class A(D):
    def showA(self):
        print("showA function")

class B(D):
    def showB(self):
        print("showB function")

class C(D):
    def showC(self):
        print("showC function")


a = A()
a.showD()
a.showA()

b = B()
b.showD()
b.showB()

c = C()
c.showD()
c.showC()
```

```python
#Method Overriding
class Parent:
  def bike(self):
    print("splender +")

class Child(Parent):
  def bike(self):
    print("HB")

obj=Child()
obj.bike()
```
